```python
df_pm25_raw = pd.read_csv("raw-data/pm25_concentration_rates.csv")
df_copd_hosp_raw = pd.read_csv("raw-data/COPD_hospital_rates.csv")
df_copd_cost_raw = pd.read_csv("raw-data/COPD_cost.csv")
df_smoking_raw = pd.read_csv("raw-data/county_health_smoking.csv")
df_poverty_raw = pd.read_csv("raw-data/census_poverty.csv")
```

```python
new_epa_df = df_pm25_raw[(df_pm25_raw["Parameter Name"] == "PM2.5 - Local Conditions")]
print(new_epa_df.shape)
new_epa_df["Pollutant Standard"].value_counts()
```

```python
new_new_epa_df = new_epa_df[(new_epa_df["Pollutant Standard"] == "PM25 Annual 2024")]
print(new_new_epa_df.shape)
print(new_new_epa_df["Event Type"].value_counts())
new_new_epa_df.head(3)
```

```python
epa_event_types_list = ["No Events", "Events Included"]
final_epa_df = new_new_epa_df[(new_new_epa_df["Event Type"].isin(epa_event_types_list))].copy()
print(final_epa_df.shape)
final_epa_df["County Name"].value_counts()
```
